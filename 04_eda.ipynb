## Hypothesis 2: Location and lot-size trade-off

Historic homes closer to central Seattle are expected to have smaller lots than comparable historic homes farther from the urban core.

The dataset does not contain golf-course locations. Distance from central Seattle is therefore used as a location proxy. Golf proximity requires external validation.

```python
seattle_lat = 47.6062
seattle_long = -122.3321
```

```python
def haversine_distance(
    lat1,
    lon1,
    lat2,
    lon2
):
    radius_miles = 3958.8

    lat1_rad = np.radians(lat1)
    lon1_rad = np.radians(lon1)
    lat2_rad = np.radians(lat2)
    lon2_rad = np.radians(lon2)

    delta_lat = lat2_rad - lat1_rad
    delta_lon = lon2_rad - lon1_rad

    a = (
        np.sin(delta_lat / 2) ** 2
        + np.cos(lat1_rad)
        * np.cos(lat2_rad)
        * np.sin(delta_lon / 2) ** 2
    )

    return (
        2
        * radius_miles
        * np.arcsin(np.sqrt(a))
    )
```

```python
h2_df = historic_non_waterfront.copy()
```

```python
h2_df["distance_to_seattle_miles"] = (
    haversine_distance(
        h2_df["lat"],
        h2_df["long"],
        seattle_lat,
        seattle_long
    )
)
```
